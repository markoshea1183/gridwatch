# Weather data Dallas

```python
# pull Dallas weather data
import pandas as pd
from datetime import date, timedelta

station = "USW00003927"  # Dallas Love Field
end_date = date.today()
start_date = end_date - timedelta(days=365)

url = (
    "https://www.ncei.noaa.gov/access/services/data/v1"
    "?dataset=daily-summaries"
    f"&stations={station}"
    f"&startDate={start_date}"
    f"&endDate={end_date}"
    "&dataTypes=TAVG,TMAX,TMIN,PRCP,SNOW,SNWD"
    "&units=standard"
    "&format=json"
)

raw = pd.read_json(url)

print(raw.columns)  # inspect what NOAA actually returned

weather = raw.rename(columns={
    "DATE": "weather_date",
    "TAVG": "temp_avg",
    "TMAX": "temp_max",
    "TMIN": "temp_min",
    "PRCP": "precipitation",
    "SNOW": "snowfall",
    "SNWD": "snow_depth"
})

# Create missing columns if NOAA did not return them
for col in ["temp_avg", "temp_max", "temp_min", "precipitation", "snowfall", "snow_depth"]:
    if col not in weather.columns:
        weather[col] = pd.NA

# Convert numeric columns first
numeric_cols = ["temp_avg", "temp_max", "temp_min", "precipitation", "snowfall", "snow_depth"]

for col in numeric_cols:
    weather[col] = pd.to_numeric(weather[col], errors="coerce")

# If TAVG is missing, approximate average temperature from max/min
weather["temp_avg"] = weather["temp_avg"].fillna(
    (weather["temp_max"] + weather["temp_min"]) / 2
)

weather["weather_date"] = pd.to_datetime(weather["weather_date"]).dt.strftime("%-m/%-d/%y")

weather[["precipitation", "snowfall", "snow_depth"]] = weather[
    ["precipitation", "snowfall", "snow_depth"]
].fillna(0)

cols = [
    "weather_date",
    "temp_avg",
    "temp_max",
    "temp_min",
    "precipitation",
    "snowfall",
    "snow_depth"
]

weather = weather[cols]

weather.to_csv("data/raw/weather/dallas_weather_past_year.csv", index=False)

weather.head()
```
